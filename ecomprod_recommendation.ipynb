```python
def update_interaction_with_product_ids(interaction_df, user_id, *product_id_lists):
    """
    Updates the interaction_df with product IDs for a specified user_id.
    
    Parameters:
    - interaction_df: DataFrame containing interactions
    - user_id: The user ID to update in interaction_df
    - product_id_lists: A variable number of product ID lists
    
    Returns:
    - Updated interaction_df
    """
    # Step 1: Combine all provided product ID lists and shuffle them
    combined_product_ids = []
    for product_id_list in product_id_lists:
        combined_product_ids.extend(product_id_list)
    
    random.shuffle(combined_product_ids)
    
    # Step 2: Filter the interactions for the specified user_id
    user_interactions = interaction_df[interaction_df['user_id'] == user_id]
    num_user_interactions = len(user_interactions)
    
    # Step 3: Adjust combined_product_ids based on the number of user interactions
    if len(combined_product_ids) < num_user_interactions:
        # If there are not enough product IDs, repeat them to fill
        combined_product_ids = (
            combined_product_ids * (num_user_interactions // len(combined_product_ids) + 1)
        )[:num_user_interactions]
    else:
        # If there are more product IDs than needed, take only the required number
        combined_product_ids = combined_product_ids[:num_user_interactions]
    
    # Step 4: Update interaction_df with the selected product IDs for the specified user_id
    interaction_df.loc[interaction_df['user_id'] == user_id, 'product_id'] = combined_product_ids
    
    return interaction_df
```

```python
def clean_product_description(description):
    # Convert to lowercase
    description = description.lower()
    
    # Remove special characters, punctuation, and extra spaces
    description = re.sub(r'[^a-z0-9.\s]', ' ', description)
    
    # Tokenize and remove stop words
    stop_words = set(stopwords.words('english'))
    tokens = description.split()
    filtered_tokens = [word for word in tokens if word not in stop_words]
    
    # Lemmatize the words to reduce them to their base form
    lemmatizer = WordNetLemmatizer()
    lemmatized_tokens = [lemmatizer.lemmatize(word) for word in filtered_tokens]
    
    # Join the cleaned tokens back into a single string
    cleaned_description = ' '.join(lemmatized_tokens)
    
    return cleaned_description
```

# Model Building

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
import pickle
```

```python
vectorizer = TfidfVectorizer(stop_words='english')
tfidf_matrix = vectorizer.fit_transform(ecom_data['tags'])
```

```python
text_similarity_matrix = cosine_similarity(tfidf_matrix)
```

```python
# This matrix captures how similar two products are based on their text features (e.g., product category, brand, description).
text_similarity_matrix
```

```python
scaler = MinMaxScaler()
normalized_prices = scaler.fit_transform(ecom_data[['product_price']].values.reshape(-1, 1))
```

```python
price_similarity_matrix = 1 - np.abs(normalized_prices - normalized_prices.T)
price_similarity_matrix = np.nan_to_num(price_similarity_matrix)
```

```python
# This matrix captures how similar two products are based on their prices.
print(price_similarity_matrix[:5, :5])  # Check a small sample of the matrix
```

```python
# expected range [0.0, 1.0]
print(f"Text Similarity Matrix Range: {text_similarity_matrix.min()} to {text_similarity_matrix.max()}")
print(f"Price Similarity Matrix Range: {price_similarity_matrix.min()} to {price_similarity_matrix.max()}")
```

```python
# The values of alpha and beta determine how much importance you give to each type of similarity when recommending products. 
# Combining text and price similarities

alpha = 0.2  # Weight for text similarity
beta = 0.8   # Weight for price similarity
combined_similarity_matrix = (alpha * text_similarity_matrix) + (beta * price_similarity_matrix)
print(f"Combined Similarity Matrix Range: {combined_similarity_matrix.min()} to {combined_similarity_matrix.max()}")
```

```python
# Calculate the average rating and number of interactions for each product
popularity_data = users_interest_data.groupby('product_id').agg(
    average_rating=('rating', 'mean'),
    num_interactions=('product_id', 'count')
).reset_index()

# Merge popularity data with the products data
ecom_data = ecom_data.merge(popularity_data, on='product_id', how='left')

# Fill NaN values (if any) with 0
ecom_data['average_rating'].fillna(0, inplace=True)
ecom_data['num_interactions'].fillna(0, inplace=True)
```

```python
# Normalize the average_rating and num_interactions using MinMaxScaler
scaler = MinMaxScaler()
ecom_data[['normalized_rating', 'normalized_interactions']] = scaler.fit_transform(
    ecom_data[['average_rating', 'num_interactions']]
)

# Calculate a combined popularity score
ecom_data['popularity_score'] = (ecom_data['normalized_rating'] + ecom_data['normalized_interactions']) / 2
```

```python
ecom_data.sample(15)
```

```python
def recommend_products_for_user(user_id, users_interest_data, products_data, similarity_matrix, top_n=10):
    # Get the list of product IDs the user is interested in
    user_interests = users_interest_data[users_interest_data['user_id'] == user_id]['product_id'].tolist()

    # Check if the user is new (i.e., no interests found)
    if not user_interests:
        print("New user detected. Providing popularity-based recommendations.")

        # Sort products based on popularity score in descending order
        popular_products = ecom_data.sort_values(by='popularity_score', ascending=False)

        # Return the top N popular products
        return popular_products['product_name'].head(top_n).tolist()
    
    # Get the indices of these products in the products_data DataFrame
    product_indices = [products_data[products_data['product_id'] == pid].index[0] for pid in user_interests]

    # Calculate the average similarity score for each product based on the user's interests
    similarity_scores = sum(similarity_matrix[idx] for idx in product_indices) / len(product_indices)

    # Sort products based on similarity scores in descending order
    product_list = sorted(list(enumerate(similarity_scores)), key=lambda x: x[1], reverse=True)

    # high threshold (e.g., 0.8), recommend products that are highly similar and low threshold (e.g., 0.1), include a wider range of products
    threshold = 0.6
    recommended_products = [
        products_data.iloc[i[0]]['product_name'] for i in product_list
        if products_data.iloc[i[0]]['product_id'] not in user_interests and i[1] > threshold
    ]

    # Return the top 5 recommended products
    print(len(recommended_products))
    return recommended_products
```

# Evaluation

```python
def split_user_interests(users_interest_data, user_id):
    user_interests = users_interest_data[users_interest_data['user_id'] == user_id]['product_id'].tolist()
    train_interests, test_interests = train_test_split(user_interests, test_size=0.3, random_state=42)
    return train_interests, test_interests
```

```python
def evaluated_recommend_products_for_user(user_id, users_interest_data, products_data, similarity_matrix):
    # Split user interests into training and testing
    train_interests, test_interests = split_user_interests(users_interest_data, user_id)
    
    # Get the indices of these products in the products_data DataFrame
    product_indices = [products_data[products_data['product_id'] == pid].index[0] for pid in train_interests]

    # Calculate the average similarity score for each product based on the user's training interests
    similarity_scores = sum(similarity_matrix[idx] for idx in product_indices) / len(product_indices)

    # Sort products based on similarity scores in descending order
    product_list = sorted(list(enumerate(similarity_scores)), key=lambda x: x[1], reverse=True)

    # Filter out products the user has already shown interest in (train set)
    # recommended_products = [
    #     products_data.iloc[i[0]]['product_id'] for i in product_list
    #     if products_data.iloc[i[0]]['product_id'] not in train_interests
    # ]
    threshold = 0.6
    recommended_products = [
        products_data.iloc[i[0]]['product_id'] for i in product_list
        if products_data.iloc[i[0]]['product_id'] not in train_interests and i[1] > threshold
    ]

    # Return the top 15 recommended product IDs
    return recommended_products[:15], test_interests
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score

def evaluate_recommendations(recommended_products, test_interests):
    y_true = [1 if pid in test_interests else 0 for pid in recommended_products]
    y_pred = [1] * len(recommended_products)

    precision = precision_score(y_true, y_pred)
    recall = recall_score(y_true, y_pred)
    f1 = f1_score(y_true, y_pred)

    print(f"Precision: {precision:.2f}")
    print(f"Recall: {recall:.2f}")
    print(f"F1-Score: {f1:.2f}")
```

```python
recommended_products, test_interests = evaluated_recommend_products_for_user(1, users_interest_data, products_data, combined_similarity_matrix)
evaluate_recommendations(recommended_products, test_interests)
```
